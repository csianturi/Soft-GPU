```python
import numpy as np

NUM_TESTS = 10000
IMG_H = 28
IMG_W = 28
IMG_SIZE = IMG_H * IMG_W

raw = np.fromfile("images.bin", dtype=np.uint8)

# Skip 16-byte header
images = raw[16:]

assert images.size == NUM_TESTS * IMG_SIZE
```

```python
def get_image(images, index):
    start = index * IMG_SIZE
    img28 = images[start:start + IMG_SIZE].reshape(28, 28).astype(np.float32)

    img32 = np.zeros((32, 32), dtype=np.float32)
    img32[2:30, 2:30] = img28

    return img32
```

```python
img = get_image(images, 0)
print(img.shape)        # (32, 32)
print(img.sum())        # non-zero
print(img[:2, :2])      # zeros (padding)
```
